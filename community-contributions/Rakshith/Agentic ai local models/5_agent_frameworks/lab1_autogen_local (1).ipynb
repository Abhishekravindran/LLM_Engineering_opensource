```python
from ag2 import AssistantAgent, UserProxyAgent
```

```python
import ag2 as autogen

HF_TOKEN = "YOUR_HUGGING_FACE_TOKEN"

config_list = [
    {
        "model": "meta-llama/Llama-3.1-8B-Instruct",
        "base_url": "https://router.huggingface.co/v1",
        "api_key": HF_TOKEN,
        "api_type": "openai",
        "price": [0, 0],
    }
]

llm_config = {
    "config_list": config_list,
    "temperature": 0.2,
    "timeout": 120,
}

assistant = autogen.AssistantAgent(
    name="assistant",
    system_message="You are a concise tutor. Explain in 3 sentences max and end with TERMINATE.",
    llm_config=llm_config,
)

user = autogen.UserProxyAgent(
    name="user",
    human_input_mode="NEVER",
    max_consecutive_auto_reply=2,
    is_termination_msg=lambda m: "TERMINATE" in (m.get("content") or ""),
    code_execution_config=False,
)

user.initiate_chat(
    assistant,
    message="Explain MCP in 3 sentences for students."
)
```
