```python
from crewai import LLM

GEMINI_API_KEY = "your Gemini API key here"  # Replace with your actual Gemini API key

llm = LLM(
    model="gemini/gemini-3.6-flash",
    api_key=GEMINI_API_KEY
)

print("CrewAI is using Gemini 3.6 Flash")
```

```python
from crewai import Agent, Task, Crew, Process

# ------------------ Agents ------------------

researcher = Agent(
    role="Researcher",
    goal="Collect 3 facts.",
    backstory="Analyst.",
    llm=llm,
)

writer = Agent(
    role="Writer",
    goal="Write a short paragraph.",
    backstory="Teacher.",
    llm=llm,
)

manager = Agent(
    role="Manager",
    goal="Delegate and deliver a clean student paragraph.",
    backstory="Project lead.",
    llm=llm,
)

# ------------------ Tasks ------------------

t_research = Task(
    description="List 3 facts about Ollama.",
    expected_output="3 bullet points.",
    agent=researcher,
)

t_write = Task(
    description="Write one short paragraph for students using the research facts about Ollama.",
    expected_output="One short paragraph.",
    agent=writer,
)

# ------------------ Sequential Process ------------------

print("==== SEQUENTIAL ====")

sequential_crew = Crew(
    agents=[researcher, writer],
    tasks=[t_research, t_write],
    process=Process.sequential,
)

sequential_result = await sequential_crew.kickoff_async()
print(sequential_result)

# ------------------ Hierarchical Process ------------------

print("\n==== HIERARCHICAL ====")

hierarchical_crew = Crew(
    agents=[researcher, writer, manager],
    tasks=[t_research, t_write],
    process=Process.hierarchical,
    manager_llm=llm,
)

hierarchical_result = await hierarchical_crew.kickoff_async()
print(hierarchical_result)
```

Hierarchical uses more tokens and can flake on tiny models — that is a reliability lesson. **Next:** tools.
