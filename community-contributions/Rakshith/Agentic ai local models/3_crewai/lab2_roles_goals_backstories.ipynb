```python
from crewai import Agent, Task, Crew

TOPIC = "Explain ReAct agents in 3 sentences."

backstories = {
    "professor": "A tenured CS professor who uses precise vocabulary.",
    "ta": "A friendly teaching assistant who uses analogies with cooking.",
    "skeptic": "An engineer who distrusts hype and always mentions failure modes.",
}

# Loop through each personality
for label, story in backstories.items():

    agent = Agent(
        role="Explainer",
        goal="Help a student understand ReAct.",
        backstory=story,
        llm=llm,
        verbose=False,
    )

    task = Task(
        description=TOPIC,
        expected_output="Exactly 3 sentences.",
        agent=agent,
    )

    crew = Crew(
        agents=[agent],
        tasks=[task],
        verbose=False,
    )

    print(f"\n==== {label.upper()} ====")

    # ✅ Use async kickoff in Colab/Jupyter
    result = await crew.kickoff_async()

    print(result)
```
