```python
import chromadb

# 1. Initialize an in-memory ChromaDB client
client = chromadb.Client()

# 2. Create a collection (uses default 'all-MiniLM-L6-v2' embeddings under the hood)
collection = client.create_collection(name="notes_collection")

# 3. Add sample documents (Chroma automatically computes the embeddings)
documents = [
    "I bought fresh apples, spinach, and bananas at the grocery store.",
    "The mechanic fixed the brakes and changed the motor oil in my car.",
    "Golden retrievers and labradors are friendly family dogs.",
    "Learned how to set up Docker containers and debug network bridges."
]

ids = ["note_1", "note_2", "note_3", "note_4"]

collection.add(
    documents=documents,
    ids=ids
)

# 4. Perform a semantic similarity query
query_text = "healthy breakfast snacks"

results = collection.query(
    query_texts=[query_text],
    n_results=2
)

# 5. Print the matches
print(f"--- Query: '{query_text}' ---\n")
for doc, distance, doc_id in zip(results["documents"][0], results["distances"][0], results["ids"][0]):
    print(f"[{doc_id}] (Distance: {distance:.4f}) -> {doc}")
```

```python
# 1. Install dependencies
!pip install -q chromadb transformers torch

import chromadb
from transformers import pipeline

# 2. Setup ChromaDB with sample data
client = chromadb.Client()
collection = client.create_collection(name="knowledge_base")

documents = [
    "I bought fresh apples, spinach, and bananas at the grocery store for healthy snacking.",
    "The mechanic fixed the brakes and changed the motor oil in my car.",
    "Golden retrievers and labradors are friendly family dogs.",
    "Learned how to set up Docker containers and debug network bridges."
]
collection.add(
    documents=documents,
    ids=[f"doc_{i}" for i in range(len(documents))]
)

# 3. Load a free, fast QA model
qa_pipeline = pipeline("text2text-generation", model="google/flan-t5-base")

# 4. User Question & Retrieval
question = "What healthy items did I buy?"
retrieval = collection.query(query_texts=[question], n_results=1)
retrieved_context = retrieval["documents"][0][0]

# 5. Build prompt and generate answer
prompt = f"""Use the following context to answer the question.
Context: {retrieved_context}
Question: {question}
Answer:"""

result = qa_pipeline(prompt, max_new_tokens=50)

print(f"Question: {question}")
print(f"Retrieved Context: {retrieved_context}")
print(f"Generated Answer: {result[0]['generated_text']}")
```

```python
import torch
from transformers import pipeline

# -------------------------------------------------------------
# 4. Load Gemma using 'text-generation'
# -------------------------------------------------------------
qa_pipeline = pipeline(
    "text-generation",
    model="google/gemma-2-2b-it",
    model_kwargs={"torch_dtype": torch.bfloat16},
    device_map="auto"
)

# -------------------------------------------------------------
# 5. Query Function with Gemma Chat Template
# -------------------------------------------------------------
def ask_document(query, n_chunks=2):
    # Retrieve relevant chunks from ChromaDB
    results = collection.query(query_texts=[query], n_results=n_chunks)
    context = "\n---\n".join(results["documents"][0])

    # Format using Gemma's chat structure
    messages = [
        {
            "role": "user",
            "content": f"""Use the following context to provide a direct, factual answer to the question.
If the answer cannot be found in the context, say "I don't know based on the provided document."

Context:
{context}

Question: {query}"""
        }
    ]

    # Generate response
    outputs = qa_pipeline(
        messages,
        max_new_tokens=150,
        do_sample=False
    )

    # Extract the generated assistant message
    generated_answer = outputs[0]["generated_text"][-1]["content"]

    print(f"\n Question: {query}")
    print(f" Generated Answer: {generated_answer}")
    print(f"\n Sourced Context:\n{context}\n" + "="*50)

# -------------------------------------------------------------
# 6. Test with Sample Queries
# -------------------------------------------------------------
ask_document("How much is the home office stipend and what does it cover?")
ask_document("Where is the annual retreat held?")
```

```python
while True:
    user_query = input("\nAsk a question (or type 'exit' to quit): ")
    if user_query.strip().lower() in ["exit", "quit", "q"]:
        print("Exiting QA session.")
        break
    if user_query.strip():
        ask_document(user_query)
```
